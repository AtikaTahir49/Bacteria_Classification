```python
import os
import shutil
from google.colab import files
import torch
from PIL import Image
import time

# Install required packages
!pip install torch torchvision openai-clip -q

# Import CLIP
import clip
import torch.nn.functional as F

# ============ STEP 1: LOAD THE MODEL ============
print("🔄 Loading CLIP model...")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model, preprocess = clip.load("ViT-B/32", device=device)
print(f"✅ Model loaded on: {device}")

# ============ STEP 2: DEFINE YOUR BACTERIA SPECIES ============
# These must match your existing folder names
bacteria_species = [
    "Staphylococcus aureus",
    "Staphylococcus saprophyticus",
    "Staphylococcus epidermidis",
    "Streptococcus pneumoniae",
    "Streptococcus pyogenes",
    "Streptococcus agalactiae"
]

# Create detailed descriptions for better classification
species_descriptions = [
    "Staphylococcus aureus - golden yellow colonies, beta hemolytic",
    "Staphylococcus saprophyticus - white cream colonies, usually gamma hemolytic",
    "Staphylococcus epidermidis - small white colonies, gamma hemolytic",
    "Streptococcus pneumoniae - mucoid colonies, alpha hemolytic, draughtsman appearance",
    "Streptococcus pyogenes - small translucent pinpoint colonies, strong beta hemolytic",
    "Streptococcus agalactiae - gray white smooth colonies, narrow beta hemolytic"
]

# Create mapping for folder names
folder_names = {species: species.replace(' ', '_') for species in bacteria_species}

# ============ STEP 3: PREPARE TEXT PROMPTS ============
print("📝 Preparing text prompts...")
text_prompts = [f"A photo of {desc}" for desc in species_descriptions]
text_tokens = clip.tokenize(text_prompts).to(device)

with torch.no_grad():
    text_features = model.encode_text(text_tokens)
    text_features = text_features / text_features.norm(dim=1, keepdim=True)

print(f"✅ Text prompts prepared for {len(text_prompts)} species")

# ============ STEP 4: UPLOAD YOUR IMAGES ============
print("\n📤 Upload your unorganized images (select all images)...")
print("(You can upload multiple files at once by Ctrl+Click or Shift+Click)")
uploaded = files.upload()

# Create a temp folder for uploaded images
temp_dir = "/content/unorganized_images"
os.makedirs(temp_dir, exist_ok=True)

# Save uploaded images
for filename in uploaded.keys():
    # Move to temp folder
    if os.path.exists(filename):
        shutil.move(filename, os.path.join(temp_dir, filename))

print(f"✅ {len(uploaded)} images uploaded to {temp_dir}")

# ============ STEP 5: CLASSIFY EACH IMAGE ============
print("\n🔬 Classifying and organizing images...")
print("="*60)

predictions = []
successful = 0
errors = 0

# Process each image
image_files = [f for f in os.listdir(temp_dir)
               if f.lower().endswith(('.jpg', '.jpeg', '.png', '.tiff', '.bmp', '.jfif'))]

if len(image_files) == 0:
    print("❌ No images found! Please upload images.")
    exit()

print(f"Found {len(image_files)} images to process\n")

for i, img_file in enumerate(image_files):
    img_path = os.path.join(temp_dir, img_file)
    print(f"  [{i+1}/{len(image_files)}] Processing: {img_file}")

    try:
        # Load and preprocess image
        image = Image.open(img_path).convert('RGB')
        image_tensor = preprocess(image).unsqueeze(0).to(device)

        # Get image features
        with torch.no_grad():
            image_features = model.encode_image(image_tensor)
            image_features = image_features / image_features.norm(dim=1, keepdim=True)

        # Calculate similarity with text prompts
        similarity = (image_features @ text_features.T).squeeze(0)
        probabilities = F.softmax(similarity, dim=0)

        # Get top prediction
        top_idx = similarity.argmax().item()
        predicted_species = bacteria_species[top_idx]
        confidence = probabilities[top_idx].item() * 100

        # Get folder name
        target_folder = folder_names[predicted_species]
        target_path = f"/content/bacteria_dataset/{target_folder}"

        # Make sure folder exists
        os.makedirs(target_path, exist_ok=True)

        # Move the image
        dest_path = os.path.join(target_path, img_file)

        # Handle duplicate filenames
        counter = 1
        while os.path.exists(dest_path):
            name, ext = os.path.splitext(img_file)
            dest_path = os.path.join(target_path, f"{name}_{counter}{ext}")
            counter += 1

        shutil.move(img_path, dest_path)
        successful += 1

        # Store prediction info
        predictions.append({
            'filename': img_file,
            'predicted': predicted_species,
            'confidence': confidence,
            'moved_to': dest_path
        })

        status = "✅" if confidence > 50 else "⚠️"
        print(f"    {status} → {predicted_species} ({confidence:.1f}% confidence)")

    except Exception as e:
        print(f"    ❌ Error: {e}")
        # Save problematic images to a separate folder
        error_dir = "/content/problematic_images"
        os.makedirs(error_dir, exist_ok=True)
        shutil.move(img_path, os.path.join(error_dir, img_file))
        errors += 1

# ============ STEP 6: VERIFY ORGANIZATION ============
print("\n" + "="*60)
print("📊 DATASET VERIFICATION")
print("="*60)

total = 0
for species in bacteria_species:
    folder_name = folder_names[species]
    path = f"/content/bacteria_dataset/{folder_name}"
    if os.path.exists(path):
        images = [f for f in os.listdir(path)
                 if f.lower().endswith(('.jpg', '.jpeg', '.png', '.tiff', '.bmp', '.jfif'))]
        print(f"  {folder_name:35} {len(images):3} images")
        total += len(images)
    else:
        print(f"  {folder_name:35} ❌ Folder missing")

print("="*60)
print(f"✅ TOTAL IMAGES ORGANIZED: {total}")

# ============ STEP 7: SUMMARY ============
print("\n📋 PROCESSING SUMMARY:")
print("="*60)
print(f"  ✅ Successfully classified: {successful} images")
print(f"  ⚠️  Low confidence (<50%): {len([p for p in predictions if p['confidence'] < 50])} images")
print(f"  ❌ Errors: {errors} images")
print("="*60)

# ============ STEP 8: SHOW LOW CONFIDENCE IMAGES ============
low_conf = [p for p in predictions if p['confidence'] < 50]
if low_conf:
    print("\n⚠️  Low confidence predictions (may need manual review):")
    print("-"*60)
    for p in low_conf:
        print(f"  • {p['filename']} → {p['predicted']} ({p['confidence']:.1f}%)")

# ============ STEP 9: SHOW ALL PREDICTIONS ============
print("\n📋 ALL PREDICTIONS:")
print("-"*60)
for p in predictions:
    print(f"  {p['filename']:30} → {p['predicted']} ({p['confidence']:.1f}%)")

# ============ STEP 10: CREATE A LOG FILE ============
with open('/content/organization_log.txt', 'w') as f:
    f.write("Bacteria Image Organization Log\n")
    f.write("="*60 + "\n\n")
    for p in predictions:
        f.write(f"Image: {p['filename']}\n")
        f.write(f"Predicted: {p['predicted']}\n")
        f.write(f"Confidence: {p['confidence']:.1f}%\n")
        f.write(f"Moved to: {p['moved_to']}\n")
        f.write("-"*60 + "\n")

print("\n✅ Log saved to: /content/organization_log.txt")
print("\n🎉 All images organized! You can now proceed with training.")
```
